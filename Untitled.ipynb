```python
import pandas as pd
import geopandas as gpd
import matplotlib.pyplot as plt
from shapely import wkt

# Load the dataset
data_path = 'UKCensus-21-11-London-21geog-data (14).csv'
data = pd.read_csv(data_path)

# Convert WKT strings to geometry
data['geometry'] = data['geometry'].apply(wkt.loads)

# Create GeoDataFrame
gdf = gpd.GeoDataFrame(data, geometry='geometry')

# Set CRS (assuming British National Grid EPSG:27700)
gdf.set_crs(epsg=27700, inplace=True)

# Calculate combined percentage of people reporting "bad" and "very bad" health
gdf['Pct_BadAndVeryBadHealth'] = ((gdf['HealthBad21'] + gdf['HealthVB21']) / gdf['TotRes21']) * 100

# Plot the map
col = 'Pct_BadAndVeryBadHealth'
fig, ax = plt.subplots(1, 1, figsize=(9, 9))

gdf.plot(
    column=col,
    cmap='plasma',           # Adjusted color map for better contrast
    scheme='quantiles',      # Optional: Quantile classification
    k=5,                     # Number of classes
    linewidth=0.8,
    legend=True,
    legend_kwds={"title": "Bad & Very Bad Health (%)", "loc": 2},
    ax=ax
)

ax.set_axis_off()  # Remove axis for cleaner look
plt.title("Percentage of People Reporting Bad & Very Bad Health (2021)", fontsize=12)
plt.show()
```

```python
gdf['Pct_BadAndVeryBadHealth'] = ((gdf['HealthBad21'] + gdf['HealthVB21']) / gdf['TotRes21']) * 100
gdf['Pct_NoQual2021'] = (gdf['QualNone21'] / gdf['TotRes21']) * 100
import statsmodels.formula.api as smf

# Define the formula for OLS regression
ols_formula = 'Pct_BadAndVeryBadHealth ~ Pct_NoQual2021'

# Fit the OLS regression model
ols_model = smf.ols(formula=ols_formula, data=gdf)
ols_fit = ols_model.fit()

# Print the summary of the regression results
print(ols_fit.summary())
```

```python
fig, ax = plt.subplots(1)
plt.scatter(gdf['Pct_NoQual2021'],gdf['Pct_BadAndVeryBadHealth'])
ax.set_ylabel('Bad And Very Bad Health (%)')
ax.set_xlabel('0 University Qualification (%)')
plt.show()
```

```python
  # Define the formula for OLS regression for Qualifications
ols_formula_qualifications = 'Pct_BadAndVeryBadHealth ~ Pct_NoQual2021'

# Fit the model for Qualifications
ols_model_qualifications = smf.ols(formula=ols_formula_qualifications, data=gdf)
ols_fit_qualifications = ols_model_qualifications.fit()
# Extract predicted values and standardized residuals for Qualifications
predicted_qualifications = ols_fit_qualifications.fittedvalues
standardized_residuals_qualifications = (ols_fit_qualifications.resid - ols_fit_qualifications.resid.mean()) / ols_fit_qualifications.resid.std()

# Plot standardized residuals against predicted values for Qualifications
plt.scatter(predicted_qualifications, standardized_residuals_qualifications, edgecolor='white')
plt.axhline(0, color='black', linestyle='dashed')
plt.xlabel('Predicted Y Values (Qualifications)')
plt.ylabel('Standardized Residuals (z)')
plt.title('Standardized Residuals vs Predicted Values (Qualifications)')
plt.show()

```

```python
import pandas as pd
from statsmodels.stats.outliers_influence import variance_inflation_factor

# Define the independent variables (X)
X = gdf[['Pct_NoQual2021', 'Pct_EthnicMinority', 'Pct_Aged65Plus', 'Pct_NoCar2021']]

# Add a constant column for intercept
X = sm.add_constant(X)
# Create an empty DataFrame to store VIF values
vif = pd.DataFrame()

# Calculate VIF for each variable
vif['VIF'] = [variance_inflation_factor(X.values, i) for i in range(X.shape[1])]
vif['variable'] = X.columns
# Display the VIF DataFrame
print(vif)
```

```python
import esda
from libpysal.weights import KNN
from splot.esda import moran_scatterplot
import matplotlib.pyplot as plt
# Create a spatial weights matrix using k-nearest neighbors
wKNN = KNN.from_dataframe(gdf, k=8)  # k=8 nearest neighbors
wKNN.transform = 'r'  # Row-standardize the weights matrix
# Calculate Moran's I for the dependent variable (Pct_BadAndVeryBadHealth)
moran_dependent = esda.Moran(gdf['Pct_BadAndVeryBadHealth'], wKNN)

# Print Moran's I and p-value
print(f"Dependent Variable Moran's I: {round(moran_dependent.I, 3)}")
print(f"Dependent Variable P-value: {round(moran_dependent.p_sim, 3)}")

# Plot Moran's I scatter plot
moran_scatterplot(moran_dependent, aspect_equal=True)
plt.title("Moran's I Scatter Plot (Dependent Variable)")
plt.show()
```

```python
# List of independent variables
independent_vars = ['Pct_NoQual2021', 'Pct_EthnicMinority', 'Pct_Aged65Plus', 'Pct_NoCar2021']

# Loop through independent variables and calculate Moran's I
for var in independent_vars:
    moran = esda.Moran(gdf[var], wKNN)
    print(f"{var} Moran's I: {round(moran.I, 3)}")
    print(f"{var} P-value: {round(moran.p_sim, 3)}")
    
    # Plot Moran's I scatter plot
    moran_scatterplot(moran, aspect_equal=True)
    plt.title(f"Moran's I Scatter Plot ({var})")
    plt.show()
```

```python
# Create a composite variable as the mean of the 4 independent variables
gdf['Composite_Factors'] = gdf[['Pct_NoQual2021', 'Pct_EthnicMinority', 'Pct_Aged65Plus', 'Pct_NoCar2021']].mean(axis=1)
# Calculate Moran's I for the composite variable
moran_composite = esda.Moran(gdf['Composite_Factors'], wKNN)

# Print Moran's I and p-value
print(f"Composite Factors Moran's I: {round(moran_composite.I, 3)}")
print(f"Composite Factors P-value: {round(moran_composite.p_sim, 3)}")

# Plot Moran's I scatter plot
moran_scatterplot(moran_composite, aspect_equal=True)
plt.title("Moran's I Scatter Plot (Composite Factors)")
plt.show()
```

```python
# Define dependent variable (y)
g_y = gdf['Pct_BadAndVeryBadHealth'].values.reshape(-1, 1)  # Replace with your dependent variable column

# Define independent variables (X)
g_X = gdf[['Pct_NoQual2021', 'Pct_EthnicMinority', 'Pct_Aged65Plus', 'Pct_NoCar2021']].values  # Replace with your independent variable columns

# Combine longitude and latitude into a list of coordinates
g_coords = list(zip(gdf['longitude'], gdf['latitude']))
```

```python
from mgwr.sel_bw import Sel_BW
from mgwr.gwr import GWR

# Select bandwidth
bw = Sel_BW(g_coords, g_y, g_X, spherical=False).search()

# Print the selected bandwidth
print(f"Optimal Bandwidth: {bw}")
```

```python
# Build the GWR model
gwr_model = GWR(g_coords, g_y, g_X, bw, fixed=False, spherical=False)

# Fit the model
gwr_results = gwr_model.fit()

# Print the summary of GWR results
print(gwr_results.summary())

```

```python
# Add local R² values to the GeoDataFrame
gdf['Local_R2'] = gwr_results.localR2

# Plot the map of local R²
import matplotlib.pyplot as plt
gdf.plot(column='Local_R2', cmap='viridis', legend=True)
plt.title('Local R² from GWR')
plt.show()
```

```python
# Add coefficients for a specific variable (e.g., Pct_NoQual2021)
gdf['Coeff_NoQual'] = gwr_results.params[:, 1]  # Replace with the index for your variable

# Plot the map of coefficients
gdf.plot(column='Coeff_NoQual', cmap='coolwarm', legend=True)
plt.title('GWR Coefficients for % with No Qualifications')
plt.show()
```
